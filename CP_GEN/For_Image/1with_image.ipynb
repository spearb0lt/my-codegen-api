````python
# import requests, json, re, html, os, sys
# from pathlib import Path

# hl = requests.get("https://my-api-mneh.onrender.com/health")
# print(hl.json())
Q_zip_path = "5th.zip"

# if (ff==0):
url = "https://my-api-mneh.onrender.com/generate"
# Q_zip_path = "PayOff.zip"
files = {"file": (Q_zip_path, open(Q_zip_path,"rb"), "application/zip")}
r = requests.post(url, files=files, timeout=1200)
r.raise_for_status()
j = r.json()
Path("gen_response.json").write_text(json.dumps(j, indent=2))
print(j)

# Save code text if present
code_text = j.get("solution")
if code_text:
    # remove code fences if present
    m = re.search(r"```(?:python)?\s*\n([\s\S]*?)```", code_text, re.IGNORECASE)
    code = m.group(1) if m else code_text
    code = html.unescape(code).replace("\r\n","\n")
    outp = Path("coding_solution_gen.py")
    outp.write_text(code, encoding="utf-8")
    print("Saved coding_solution_gen.py")
else:
    print("No solution text returned. Use download endpoint with solution_id.")

#DOWNLOAD

solution_id = j.get("solution_id")
solution_path = j.get("solution_path")
# "solution_path": "solutions/7976324c-bc1e-4331-8a6f-2b2e24ac57ab/coding_solution.py",
print(f"solution_id: {solution_id}")
print(f"solution_path: {solution_path}")    
# "solution_path": "solutions/7976324c-bc1e-4331-8a6f-2b2e24ac57ab/coding_solution.py",
durl=f"https://my-api-mneh.onrender.com/download/{solution_id}/coding_solution.py"
r = requests.get(durl)
r.raise_for_status()
open("coding_solution_downloaded.py","wb").write(r.content)
````

````python
# CHECK LOCAL RUNNING


# import subprocess
# import sys
# import traceback

# === Configurable parameters ===
input_file = "test_input.txt"





output_file = "sample_out.txt"
error_file = "sample_err.txt"   # will store stderr / TLE message if any
time_limit = 180  # seconds (adjust as needed)
program = "coding_solution_gen.py"

error_message = None

# Use context managers so files are closed reliably
with open(input_file, "r") as inf, open(output_file, "w") as outf:
    try:
        result = subprocess.run(
            [sys.executable, program],   # more robust than "python"
            stdin=inf,                  # pass the file object as stdin
            stdout=outf,
            stderr=subprocess.PIPE,
            timeout=time_limit,
            text=True
        )

        # non-zero exit code -> runtime error captured in stderr
        if result.returncode != 0:
            error_message = result.stderr.strip()
            # Save the error to a file (optional)
            with open(error_file, "w") as ef:
                ef.write(error_message)
            print("Error occurred:\n", error_message)
        else:
            error_message = None
            print("✅ Output successfully written to", output_file)

    except subprocess.TimeoutExpired:
        error_message = f"Time Limit Exceeded ({time_limit} seconds)"
        with open(error_file, "w") as ef:
            ef.write(error_message)
        print(error_message)

    except Exception as e:
        # In case of other errors (e.g., file missing, permission issues)
        tb = traceback.format_exc()
        error_message = f"Runtime Error: {e}\n{tb}"
        with open(error_file, "w") as ef:
            ef.write(error_message)
        print(error_message)

# `error_message` variable now holds None (if success) or the error/TLE text.
````

````python
#DELETE SOLUTION

import requests

def delete_solution(server_base, solution_id, timeout=30):
    url = server_base.rstrip('/') + f"/cleanup/{solution_id}"
    try:
        r = requests.delete(url, timeout=timeout)
    except Exception as e:
        print("Request failed:", e)
        return 2
    if r.status_code == 200:
        print("✅ Deleted solution:", solution_id)
        print("Response:", r.json())
        return 0
    else:
        print("❌ Delete failed:", r.status_code, r.text)
        return 1

# Example usage:
delete_solution("https://my-api-mneh.onrender.com",solution_id)
````
